```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("datasets/student_marks.csv")
# df.info()
# df.describe()
# df.isnull().sum()

# Features k ho(input), hamilai chaeyeyo label(output/predicted value) K ho
X = df[["StudyHours"]] #features(input)
Y = df["FinalMarks"] #label

# plt.scatter(X,Y)
# plt.xlabel("Study Hours")
# plt.ylabel("Final Marks")
# plt.title("Study hours vs Final Marks")

# Y.shape
X = X.to_numpy()
Y = Y.to_numpy()

#machine learning
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
model = LinearRegression()
model.fit(X_train,Y_train)

predicted_Y = model.predict(X_test)


comparision = pd.DataFrame({
    "Actual" : Y_test, 
    "Predicted" : predicted_Y
})
comparision
# X_test

new_students = np.array([
    [2],
    [4],
    [6],
    [8],
    [22]
])
marks_prediction = model.predict(new_students)
marks_prediction
```
